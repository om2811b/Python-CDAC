Q. Banking System Using Dictionary and Objects


```python
class Account:
  def __init__(self, acc_no, name, balance):
    self.acc_no = acc_no
    self.name = name
    self.balance = balance

  def deposit(self, amount):
    if amount<=0:
      return False
    self.balance += amount
    return True

  def withdraw(self, amount):
    if amount<=0:
      return "Invalid"
    if self.balance-amount < 3000:
      return "minimum"
    self.balance -= amount
    return "success"

bank = {}
next_acc_no = 1001

def open_acc():
  global next_acc_no
  name = input("Enter Account holder name: ")

  try:
    initial_dep = float(input("Enter Initial Deposit: ₹"))
  except ValueError:
    print("Invalid amount!")
    return

  if initial_dep < 5000:
    print("Account cannot be opened.")
    print("Minimum initial deposit is ₹5000.")
    return

  account = Account(next_acc_no, name, initial_dep)
  bank[next_acc_no] = account
  print("Account created successfully!")
  print(f"Account Number: {next_acc_no}")
  print(f"Current Balance: ₹{initial_dep}")

  next_acc_no += 1

def dep_money():
  try:
    acc_no = int(input("Enter Account Number: "))
  except ValueError:
    print("Invalid account number!")
    return

  if acc_no not in bank:
    print("Source account does not exist.")
    return

  try:
    amount = float(input("Enter Deposit Amount: ₹"))

  except ValueError:
    print("Invalid amount!")
    return

  account = bank[acc_no]
  if account.deposit(amount):
    print(f"₹{amount} deposited successfully.")
    print(f"Current Balance: ₹{account.balance}")
  else:
    print("Invalid amount!")
    print("Deposit amount must be positive.")

def withdraw_money():
  try:
    acc_no = int(input("Enter Account Number: "))
  except ValueError:
    print("Invalid account number!")
    return

  if acc_no not in bank:
    print("Source account does not exist.")
    return

  try:
    amount = float(input("Enter Withdrawal Amount: ₹"))

  except ValueError:
    print("Invalid amount!")
    return

  account = bank[acc_no]

  res = account.withdraw(amount)

  if res == "success":
    print("Withdrawal successful!")
    print(f"Amount Withdrawn: ₹{amount}")
    print(f"Remaining Balance: ₹{account.balance}")

  elif res == "invalid":
    print("Invalid amount!")
    print("Withdrawal amount must be positive.")

  else:
    print("Withdrawal not allowed!")
    print("Minimum balance of ₹3000 must be maintained.")
    print(f"Current Source Balance: ₹{account.balance}")


def transfer_money():
  try:
    source_no = int(input("Enter Source Account Number: "))
  except ValueError:
    print("Invalid account number!")
    return

  if source_no not in bank:
    print("Source account does not exist.")
    print("Transfer cancelled.")
    return

  try:
    destination_no = int(input("Enter Destination Account Number: "))
  except ValueError:
    print("Invalid account number!")
    return

  if destination_no not in bank:
    print("Destination Account does not exist.")
    print("Transfer cancelled.")
    return

  try:
    amount = float(input("Enter Transfer Amount: ₹"))

  except ValueError:
    print("Invalid amount!")
    return

  if amount <= 0:
    print("Invalid amount!")
    print("Transfer amount must be positive.")
    return

  source_acc = bank[source_no]
  destination_acc = bank[destination_no]

  if source_acc.balance - amount <3000:
    print("Transfer not allowed!")
    print("Minimum balance of ₹3000 must be maintained.")
    print(f"Current Source Balance: "
          f"₹{source_acc.balance}"
    )
    return

  source_acc.balance -= amount
  destination_acc.balance += amount
  print("Transfer successful!")
  print(f"₹{amount} transferred successfully.")
  print(f"Destination Balance : ₹{destination_acc.balance}")

def check_balance():

    try:
        account_no = int(input("Enter Account Number: "))

    except ValueError:
        print("Invalid account number!")
        return

    if account_no not in bank:
        print("Account does not exist.")
        return

    account = bank[account_no]
    print("\n----- ACCOUNT DETAILS -----")
    print(f"Account Number : {account.acc_no}")
    print(f"Account Holder : {account.name}")
    print(f"Balance        : ₹{account.balance}")


while True:
    print("\n===== BANKING SYSTEM =====")
    print("1. Open New Account")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Account-to-Account Transfer")
    print("5. Check Balance")
    print("6. Exit")

    try:
        choice = int(input("\nEnter your choice: "))

    except ValueError:
        print("Invalid choice! Please enter a number.")
        continue

    if choice == 1:
        open_acc()
    elif choice == 2:
        dep_money()
    elif choice == 3:
        withdraw_money()
    elif choice == 4:
        transfer_money()
    elif choice == 5:
        check_balance()
    elif choice == 6:
        print("\nThank you for using our Banking System!")
        break

    else:
        print("Invalid choice! Please select between 1 and 6.")
```
