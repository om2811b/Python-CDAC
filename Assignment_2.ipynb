**Problem Definition:** Menu-Driven Café Billing System


```python
items = {
    101: ["Tea", 20.0],
    102: ["Coffee", 40.0],
    103: ["Maggi", 60.0],
    104: ["Bun Maska", 50.0],
    105: ["Water", 20.0]
}

bill = {}

while True:
  print("========== CAFE BILLING SYSTEM ==========")
  print("101. Tea          ₹20.00")
  print("102. Coffee       ₹40.00")
  print("103. Maggi        ₹60.00")
  print("104. Bun Maska    ₹50.00")
  print("105. Water        ₹20.00")
  print("0. Generate Bill")
  print("=========================================")

  choice = int(input("Enter item number: "))

  if choice == 0:
    break

  elif choice in items:
    qty = int(input("Enter quantity: "))

    if qty <= 0:
      print("Invalid quantity.")
      print("Please enter a quantity greater than 0.")

    else:
      name = items[choice][0]
      cost = items[choice][1]

      if choice in bill:
        bill[choice][2] = bill[choice][2] + qty
        bill[choice][3] = bill[choice][1] * bill[choice][2]
        print(f"{name}, quantity updated.")

      else:
        total = cost * qty
        bill[choice] = [name, cost, qty, total]
        print("{name} added to bill.")

  else:
      print("Invalid item number.")
      print("Please select a valid item.")

print("================ FINAL BILL ================")
print(f"{"Item": <15}{"Qty": <10}{"Rate": <12}{"Total"}")
print("------------------------------------------------")

sub_total = 0

for item in bill:
  name = bill[item][0]
  cost = bill[item][1]
  qty = bill[item][2]
  total = bill[item][3]

  print(f"{name:<15}{qty:<10}₹{cost:<11.2f}₹{total:.2f}")

  sub_total = sub_total + total

print("------------------------------------------------")

gst = sub_total * 2.5/100
cgst = sub_total * 2.5/100

total_tax = gst + cgst
total_payable = sub_total + total_tax

print(f"Subtotal:       ₹{sub_total:.2f}")
print()
print(f"GST @ 2.5%:     ₹{gst:.2f}")
print(f"CGST @ 2.5%:    ₹{cgst:.2f}")

print("------------------------------------------------")

print(f"Total Tax:       ₹{total_tax:.2f}")
print()
print(f"TOTAL PAYABLE:    ₹{total_payable:.2f}")

print("------------------------------------------------")

print("           Thank You!")
print("          Visit Again!")
```
