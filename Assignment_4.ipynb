Q1. Reverse a String


```python
def reverse_string(s):
  return s[::-1]
s = "Hello World!"
print(reverse_string(s))
```

Q2. Check Palindrome (Ignore Case and Spaces)


```python
def is_palindrome(s):
  s = "".join(ch.lower() for ch in s if ch.isalnum())
  return s == s[::-1]
s = "A man a plan a canal Panama"
print(is_palindrome(s))
```

Q3. Count Vowels and Consonants


```python
def count_letters(s):
  vowels = set("aeiou")
  res = {"vowels": 0, "consonants": 0}

  for char in s.lower():
    if char.isalpha():
      if char in vowels:
        res["vowels"] += 1
      else:
        res["consonants"] += 1
  return res
  s = "Hello World 2026!"
print(count_letters(s))
```

Q4. Character Frequency Map


```python
def char_frequency(s):
  freq = {}
  for ch in s:
    freq[ch] = freq.get(ch, 0) + 1
  return freq
s = "placement"
print(char_frequency(s))
```

Q5. First Non-Repeating Character


```python
def first_unique_char(s):
  freq = {}

  for ch in s:
    freq[ch] = freq.get(ch, 0) + 1
  for ch in s:
    if freq[ch] == 1:
      return ch
  return None
print(first_unique_char("swiss"))
```

Q6. Check Anagrams


```python
def are_anagrams(s1, s2):
  s1 =  "".join(ch.lower() for ch in s1 if ch.isalnum())
  s2 =  "".join(ch.lower() for ch in s2 if ch.isalnum())

  return sorted(s1) == sorted(s2)
print(are_anagrams("Listen", "Silent"))
```

Q7. Remove Duplicates from a List (Preserve Order)


```python
def unique_preserve_order(items):
  seen = set()
  res = []
  for item in items:
    if item not in seen:
      seen.add(item)
      res.append(item)
  return res
items = [4, 5, 4, 1, 5, 2, 1, 3]
print(unique_preserve_order(items))
```

Q8. Second Largest Distinct Number


```python
def second_largest(nums):
  first = second = None

  for n in nums:
    if first is None or n > first:
      second = first
      first = n
    elif n!=first and (second is None or n > second):
      second = n
  return second

nums = [10, 20, 4, 20, 18, 10]
print(second_largest(nums))
```

Q9. Rotate a List by k Positions


```python
def rotate_right(nums, k):
  if not nums:
    return []
  k %= len(nums)
  if k == 0:
    return nums
  return nums[-k:] + nums[:-k]

print(rotate_right([1, 2, 3, 4, 5], 2))
```

Q10. Pair With Given Sum (Two Sum — Values)


```python
def pair_with_sum(nums, target):
  seen = set()
  for n in nums:
    need = target - n
    if need in seen:
      return(need, n)
    seen.add(n)
  return None
nums = [2, 7, 11, 15]
target = 13
print(pair_with_sum(nums, target))
```

Q11. Word Frequency in a Sentence


```python
def word_frequency(sentence):
  freq = {}
  words = sentence.lower().split()

  for word in words:
    word = "".join(ch for ch in word if ch.isalnum())
    if word:
      freq[word] = freq.get(word, 0) + 1
  return freq
print(word_frequency("Python is easy, python is powerful."))
```

Q12. Group Anagrams


```python
def group_anagrams(words):
  groups = {}
  for word in words:
    key = "".join(sorted(word))
    groups.setdefault(key, []).append(word)
  return list(groups.values())

words = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(group_anagrams(words))
```

Q13. Flatten a Nested List (One Level or Deep)


```python
def flatten(items):
  res = []
  for n in items:
    if isinstance(n, list):
      res.extend(flatten(n))
    else:
      res.append(n)
  return res

print(flatten([1, [2, [3, 4], 5], 6]))
```

Q14. Merge Two Dictionaries (Sum Common Keys)


```python
def merge_sum(d1, d2):
  res = dict(d1)
  for key, value in d2.items():
    res[key]= res.get(key,0)+value
  return res
d1 = {'a': 10, 'b': 20, 'c': 30}
d2 = {'b': 5, 'c': 15, 'd': 40}
print(merge_sum(d1, d2))
```

Q15. Students Above Average (Dict of Lists)


```python
def toppers(scores):
  all_marks=[m for marks in scores.values() for m in marks]
  if not all_marks:
    return[]
  class_avg = sum(all_marks)/len(all_marks)
  names = []
  for name, marks in scores.items():
    if marks and (sum(marks)/len(marks)) > class_avg:
      names.append(name)
  return sorted(names)

print(toppers({
    "Asha": [80, 90, 70],
    "Rohan": [60, 65, 55],
    "Meera": [90, 95, 85],
    "Dev": [70, 70, 70],
}))
```
