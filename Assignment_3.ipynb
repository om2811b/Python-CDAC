Q.1 Build a program that takes a paragraph as input and performs multiple text-analysis operations.

```python
paragraph=input("Enter Paragraph: ").lower()

vowels=consonants=digits=special=sentences=char_without_space=0
total_words=0
freq = {}
longest_word = ""
length = 0

curr_word = ""

for ch in paragraph:
  if ch != " ":
     char_without_space += 1

  if ch.isalpha():
    if ch in "aeiou":
      vowels += 1
    else:
      consonants += 1

  elif ch.isdigit():
    digits += 1

  elif ch != " ":
    special += 1

  if ch in ".!?":
    sentences += 1

  if ch.isalnum():
    curr_word += ch

  else:
    if curr_word != "":
      word = curr_word
      total_words += 1

      if word in freq:
        freq[word] += 1
      else:
        freq[word] = 1

      if len(word) > length:
        longest_word = word
        length = len(word)

      curr_word = ""

if curr_word != "":
  word = curr_word
  total_words += 1

  if word in freq:
    freq[word] += 1
  else:
    freq[word] = 1

  if len(word) > length:
        longest_word = word
        length = len(word)

if freq:
  most_frequent_word = max(freq, key=freq.get)
  most_frequent_freq = freq[most_frequent_word]
else:
  most_frequent_word = ""
  most_frequent_freq = 0

print(f"Total Characters : {len(paragraph)}")
print(f"Characters Without Spaces: {char_without_space}")
print(f"Total Words: {total_words}")
print(f"Total Sentences: {sentences}")
print(f"Vowels: {vowels}")
print(f"Consonants: {consonants}")
print(f"Digits: {digits}")
print(f"Special Characters: {special}")
print(f"Most Frequent Word: {most_frequent_word}")
print(f"Frequency: {most_frequent_freq}")
print(f"Longest Word: {longest_word}")
print(f"Length: {length}")
```
