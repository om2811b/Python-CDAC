Question 1. (Menu-Driven Programming) Write a menu-driven program for a small canteen


```python
while True:
  print("---Canteen Menu---")
  print("1. Tea → ₹10")
  print("2. Coffee → ₹15")
  print("3. Water → ₹5")
  print("4. Bunmuska → ₹20")

  choice=int(input("Enter your choice:"))

  match choice:
    case 1:
      qty=int(input("Enter quantity: "))
      rate=10
      print("----- Bill -----")
      print(f"Item: Tea")
      print(f"Quantity: {qty}")
      print(f"Rate: ₹{rate}")
      print(f"Total Amount: ₹{rate*qty}")
      print("Thank you! Visit again.")
      break

    case 2:
      qty=int(input("Enter quantity: "))
      rate=15
      print("----- Bill -----")
      print(f"Item: Coffee")
      print(f"Quantity: {qty}")
      print(f"Rate: ₹{rate}")
      print(f"Total Amount: ₹{rate*qty}")
      print("Thank you! Visit again.")
      break

    case 3:
      qty=int(input("Enter quantity: "))
      rate=5
      print("----- Bill -----")
      print(f"Item: Water")
      print(f"Quantity: {qty}")
      print(f"Rate: ₹{rate}")
      print(f"Total Amount: ₹{rate*qty}")
      print("Thank you! Visit again.")
      break

    case 4:
      qty=int(input("Enter quantity: "))
      rate=20
      print("----- Bill -----")
      print(f"Item: Bunmuska")
      print(f"Quantity: {qty}")
      print(f"Rate: ₹{rate}")
      print(f"Total Amount: ₹{rate*qty}")
      print("Thank you! Visit again.")
      break

    case _:
      print("Invalid choice. Please select a valid menu item.")
```

Question 2. (Pattern Printing) Write a program that takes a positive integer n and prints an alphabet triangle in the following format (using uppercase letters):

```python
n = int(input("Enter number of rows to print: "))
for i in range(1, n + 1):
  for j in range(i):
    print(chr(ord('A') + 2 * j), end=" ")
  print()
```

Question 3.(Series Generation & Summation) Write a program that accepts three inputs: first term (a), common difference (d), and number of terms (n). Generate the Arithmetic Progression and also print its sum.


```python
a=int(input("Enter first term: "))
d=int(input("Enter common difference: "))
n=int(input("Enter number of terms: "))

total = 0

print("AP Series:", end=" ")

for i in range(n):
    term = a + i * d
    print(term, end=" ")
    total += term

print()
print(f"Sum = {total}")
```

Question 4.(Digit Extraction) Write a program that takes a positive integer as input and counts how many even digits and how many odd digits are present in it.

```python
no = int(input("Enter a positive integer: "))
even = odd = 0
while no>0:
  digit = no%10
  if digit%2 == 0:
    even += 1
  else:
    odd += 1

  no = no // 10
print(f"Even digits: {even}")
print(f"Odd digits: {odd}" )
```

Question 5.(Digit Extraction) Write a program that accepts a positive integer and calculates:
• The sum of all even digits in the number
• The sum of all odd digits in the number

```python
no = int(input("Enter a positive integer: "))
even = odd = 0
while no>0:
  digit = no%10

  if digit%2 == 0:
    even += digit

  else:
    odd += digit

  no = no // 10
print(f"Sum of even digits= {even}")
print(f"Sum of odd digits= {odd}" )
```
