1. 🎓 Student Result Comparator


```python
class Student:
    def __init__(self, name, rollno, marks):
        self.name = name
        self.rollno = rollno
        self.marks = marks

    def __gt__(self, other):
        return self.marks > other.marks

    def __lt__(self, other):
        return self.marks < other.marks

    def __eq__(self, other):
        return self.marks == other.marks

name1 = input("Enter student name: ")
rollno1 = int(input("Enter student roll number: "))
marks1 = float(input("Enter student marks: "))
student1 = Student(name1, rollno1, marks1)

name2 = input("Enter student name: ")
rollno2 = int(input("Enter student roll number: "))
marks2 = float(input("Enter student marks: "))
student2 = Student(name2, rollno2, marks2)

print("student 1: ")
print(f"Name: {student1.name}")
print(f"Roll: {student1.rollno}")
print(f"Marks: {student1.marks}")

print("student 2: ")
print(f"Name: {student2.name}")
print(f"Roll: {student2.rollno}")
print(f"Marks: {student2.marks}")

if student1 > student2:
    print(f"Result: {student1.name} has higher marks.")

elif student1 < student2:
    print(f"Result: {student2.name} has higher marks.")

else:
    print(f"Both students have same marks.")
```

2. 🚗 Vehicle Showroom


```python
class Vehicle :
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

class Car(Vehicle ):
  def display_details(self):
    print("Car")
    print(f"Brand: {self.brand}")
    print(f"Model: {self.model}")

class Bike(Vehicle ):
  def display_details(self):
    print("Bike")
    print(f"Brand: {self.brand}")
    print(f"Model: {self.model}")

class Truck(Vehicle ):
  def display_details(self):
    print("Truck")
    print(f"Brand: {self.brand}")
    print(f"Model: {self.model}")

car = Car("Honda", "City")
bike = Bike("RE", "Bullet")
truck = Truck("Ashok", "Leyland")

vehicles = [car, bike, truck]

print("----- Vehicle Details -----")

for vehicle in vehicles:
  vehicle.display_details()
  print()
```

3. 🍕 Pizza Bill Calculator


```python
class Pizza:
  def __init__(self, base_price):
    self.base_price = base_price

  def calculate_price(self):
    return self.base_price

class RegularPizza(Pizza):
  def calculate_price(self):
    return self.base_price

class CheesePizza(Pizza):
  def __init__(self, base_price, cheese_cost):
    super().__init__(base_price)
    self.cheese_cost = cheese_cost

  def calculate_price(self):
    return self.base_price + self.cheese_cost

class PremiumPizza(Pizza):
  def __init__(self, base_price, toppings_cost):
    super().__init__(base_price)
    self.toppings_cost = toppings_cost

  def calculate_price(self):
    return self.base_price + self.toppings_cost

regular = RegularPizza(200)
cheese = CheesePizza(200, 80)
premium = PremiumPizza(200, 150)

pizzas = [regular, cheese, premium]

print(f"Regular Pizza Price: {regular.calculate_price()}")
print(f"Cheese Pizza Price: {cheese.calculate_price()}")
print(f"Premium Pizza Price: {premium.calculate_price()}")
```

4. 📦 Product Price Comparison


```python
class Product:
  def __init__(self, name, id, price):
    self.name = name
    self.id = id
    self.price = price

  def __gt__(self, other):
    return self.price > other.price

  def __lt__(self, other):
    return self.price < other.price

  def __eq__(self, other):
    return self.price == other.price

name1 = input("Enter Product 1 name: ")
id1 = input("Enter Product 1 ID: ")
price1 = input("Enter Product 1 Price: ")
product1 = Product(name1, id1, price1)

print(f"Product 1: {product1.name}")
print(f"Price: {product1.price}")

name2 = input("Enter Product 2 name: ")
id2 = input("Enter Product 2 ID: ")
price2 = input("Enter Product 2 Price: ")
product2 = Product(name2, id2, price2)

print(f"Product 2: {product2.name}")
print(f"Price: {product2.price}")

if product1 > product2:
  print(f"Result: {product1.name} is more expensive.")
  print(f"{product2.name} is cheaper.")

elif product1 < product2:
  print(f"Result: {product2.name} is more expensive.")
  print(f"{product1.name} is cheaper.")

else:
  print("Both products are of same price.")
```

5. 🏦 Digital Wallet


```python
class DigitalWallet:
  def __init__(self, owner, wallet_id, balance):
    self.owner = owner
    self.wallet_id = wallet_id
    self.__balance = balance   # Private variable

  def add_money(self, amount):
    if amount > 0:
      self.__balance += amount
      print(f"Money Added: {amount}")
    else:
      print("Invalid amount.")

  def spend_money(self, amount):
    if amount <= 0:
      print("Invalid amount.")

    elif amount > self.__balance:
      print("Insufficient balance.")

    else:
      self.__balance -= amount
      print(f"Money Spent: {amount}")

  def check_balance(self):
    return self.__balance

owner = input("Enter Owner Name: ")
wallet_id = input("Enter Wallet ID: ")
initial_balance = float(input("Enter Initial Balance: "))

wallet = DigitalWallet(owner, wallet_id, initial_balance)

print(f"Owner: {wallet.owner}")
print(f"Initial Balance: {wallet.check_balance()}")

add_amount = float(input("\nEnter Amount to Add: "))
wallet.add_money(add_amount)

print("Current Balance:", wallet.check_balance())

spend_amount = float(input("\nEnter Amount to Spend: "))
wallet.spend_money(spend_amount)

print("Final Balance:", wallet.check_balance())
```

6. 🎮 Game Characters


```python
class Character:
    def __init__(self, name, health):
        self.name = name
        self.health = health

class Warrior(Character):
    def attack(self):
        print(f"{self.name} attacks using a sword.")

class Wizard(Character):
    def attack(self):
        print(f"{self.name} attacks using magic.")

class Archer(Character):
    def attack(self):
        print(f"{self.name} attacks using a bow.")

warrior = Warrior("Thor", 100)
wizard = Wizard("Merlin", 80)
archer = Archer("Robin", 90)

characters = [warrior, wizard, archer]

for character in characters:
    character.attack()
```

7. 🧮 Smart Calculator Objects


```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def __sub__(self, other):
        return Number(self.value - other.value)

num1 = Number(int(input("Enter Number 1: ")))
num2 = Number(int(input("Enter Number 2: ")))

addition = num1 + num2
subtraction = num1 - num2

print(f"Number 1: {num1.value}")
print(f"Number 2: {num2.value}")

print(f"Addition: {addition.value}")
print(f"Subtraction: {subtraction.value}")
```

8. 💳 Payment Gateway


```python
class Payment:
    def __init__(self, amount):
        self.amount = amount

    def process_payment(self):
        pass

class UPI(Payment):
    def __init__(self, amount, upi_id):
        super().__init__(amount)
        self.upi_id = upi_id

    def process_payment(self):
        print(f"Payment processed successfully using UPI.")

class CreditCard(Payment):
    def __init__(self, amount, card_number):
        super().__init__(amount)
        self.card_number = card_number

    def process_payment(self):
        print("Payment processed successfully using Credit Card.")

class NetBanking(Payment):
    def __init__(self, amount, bank_name):
        super().__init__(amount)
        self.bank_name = bank_name

    def process_payment(self):
        print("Payment processed successfully using Net Banking.")

amount = int(input("Enter Payment Amount: "))
upi_id = input("Enter UPI ID: ")

payment = UPI(amount, upi_id)

print(f"Payment Amount: {payment.amount}")
print(f"Payment Method: UPI")
print(f"UPI ID: {payment.upi_id}")

payment.process_payment()
```
